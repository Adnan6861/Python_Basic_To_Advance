# Problem 1

 condition statement
 If a number is divisible by both 3 and 5 → print "FizzBuzz"

 If only divisible by 3 → print "Fizz"

If only divisible by 5 → print "Buzz"

 Otherwise → print the number

if used number / 3 and number /5   ---> it also give value in number always --> if values in number not Zero -->Result is True 

so They Execute it

```python
number=int(input('Enter Number :'))
if number%3==0 and number%5==0:
    print('FIzzBuzz')
elif number%3==0:
    print('Fizz')
elif number%5==0:
    print('Buzz')
else:
    print(number)
```
