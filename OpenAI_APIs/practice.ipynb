```python
import os
from dotenv import load_dotenv
load_dotenv()                        # save the Env File  than load this Function
def key_name_load(key_name):
    try:
     key=os.getenv(key_name)   # get key From env and also Store in  key variable
     if key:
        print(f"{key_name} Load Succesfully")   # all this code for validation value are Presented in key or not
        return key
     else :
        print(f'Key Not Found')
        return None
    except Exception as e :
        print(e)
open_ai_key=key_name_load('OPENAI_API_KEY')
```

```python
from openai import OpenAI
client=OpenAI(api_key=open_ai_key)
if client:
    print('SucessFull Executed')
else:
    print('Not Loaded')
```

```python
def get_ai_asistence(prompt):
    try:
        response=client.chat.completions.create(
            model='gpt-4.1-mini',
              messages=[
                {"role": "system", "content": "You are a helpful assistant."},
                {"role": "user", "content": prompt}
            ],
            max_tokens=150,
            # Control randomness (0 = deterministic, 1 = very random)
            temperature=0.7
        )
        return response.choices.message.content
    except Exception as e:
        print('Error',e)
  
    

prompt="My Name is Adnan and Your Name"
get_ai_asistence(prompt)

```
